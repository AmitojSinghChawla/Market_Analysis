```python
import datetime

import pandas as pd

from db import get_connected
```

```python
def establish_connection():
    try:
        conn=get_connected()
        return conn
    except Exception as e:
        print(e)
```

```python
REFERENCE_WINDOW= 30

DRIFT_FEATURES = [
    "daily_return",
    "ma_7",
    "ma_21",
    "ma_ratio",
    "rsi",
    "volatility_7",
    "volume_change",
    "macd",
    "stochastic",
]
```

```python
def get_data(conn) -> pd.DataFrame:

    cursor = conn.cursor()

    query="SELECT date,daily_return,ma_7,ma_21, ma_ratio, rsi,volatility_7,volume_change,macd,stochastic FROM features ORDER BY date"

    df=pd.read_sql(query,conn)

    return df
```

```python
conn=establish_connection()
df=get_data(conn)
```

```python
def separate_data(data:pd.DataFrame):

    cutoff_date= datetime.datetime.today() - datetime.timedelta(days=REFERENCE_WINDOW)

    current=data[data['date']>=cutoff_date]
    refrence= data[data['date']<cutoff_date]






    return current , refrence


```

```python
current, refrence = separate_data(df)
```
